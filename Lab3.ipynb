```python
class RoadNetwork:
    def __init__(self, connections):
        self.adj = {}
        self.cities = set()
        for u, v in connections:
            if u not in self.adj: self.adj[u] = []
            if v not in self.adj: self.adj[v] = []
            self.adj[u].append(v)
            self.adj[v].append(u)
            self.cities.add(u)
            self.cities.add(v)

    def has_route(self, start, end):
        """Checks if a route exists between start and end using DFS."""
        if start not in self.cities or end not in self.cities:
            return False

        visited = set()
        def dfs(node):
            if node == end:
                return True
            visited.add(node)
            for neighbor in self.adj.get(node, []):
                if neighbor not in visited:
                    if dfs(neighbor):
                        return True
            return False

        return dfs(start)

    def find_connected_components(self):
        """Finds all separate groups of connected cities using DFS."""
        visited = set()
        components = []

        for city in self.cities:
            if city not in visited:
                component = []
                # Helper DFS to collect all nodes in this component
                stack = [city]
                while stack:
                    node = stack.pop()
                    if node not in visited:
                        visited.add(node)
                        component.append(node)
                        stack.extend(self.adj.get(node, []))
                components.append(component)
        return components

# --- Example Usage ---
# Define a road network (pairs of connected cities)
roads = [
    ("New York", "Boston"),
    ("Boston", "Chicago"),
    ("San Francisco", "Los Angeles"),
    ("Seattle", "Portland"),
    ("Chicago", "Detroit")
]

network = RoadNetwork(roads)

# 1. Check routes
city_a, city_b = "New York", "Detroit"
route_exists = network.has_route(city_a, city_b)
print(f"Route exists between {city_a} and {city_b}: {route_exists}")

city_c, city_d = "New York", "Los Angeles"
route_exists_2 = network.has_route(city_c, city_d)
print(f"Route exists between {city_c} and {city_d}: {route_exists_2}")

# 2. Find connected components
all_groups = network.find_connected_components()
print(f"\nConnected Components (Groups of cities):")
for i, group in enumerate(all_groups, 1):
    print(f"Group {i}: {group}")
```
