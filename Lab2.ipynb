```python
from collections import deque

def water_jug_bfs(cap_a, cap_b, target):
    # Initial state: both jugs are empty (0, 0)
    # Queue stores: (current_a, current_b, path_to_this_state)
    queue = deque([(0, 0, [])])
    visited = set([(0, 0)])

    while queue:
        curr_a, curr_b, path = queue.popleft()

        # Check if either jug has the target amount
        if curr_a == target or curr_b == target:
            return path + [(curr_a, curr_b)]

        # Possible next states
        states = [
            (cap_a, curr_b, "Fill Jug A"),       # Fill A
            (curr_a, cap_b, "Fill Jug B"),       # Fill B
            (0, curr_b, "Empty Jug A"),         # Empty A
            (curr_a, 0, "Empty Jug B"),         # Empty B
            # Pour A -> B
            (curr_a - min(curr_a, cap_b - curr_b), curr_b + min(curr_a, cap_b - curr_b), "Pour A to B"),
            # Pour B -> A
            (curr_a + min(curr_b, cap_a - curr_a), curr_b - min(curr_b, cap_a - curr_a), "Pour B to A")
        ]

        for next_a, next_b, action in states:
            if (next_a, next_b) not in visited:
                visited.add((next_a, next_b))
                queue.append((next_a, next_b, path + [(curr_a, curr_b, action)]))

    return None

# Solve for 4L and 3L jugs to get 2L
solution = water_jug_bfs(4, 3, 2)

if solution:
    print(f"Minimum sequence of steps to reach 2L:")
    for i, step in enumerate(solution[:-1]):
        print(f"{i+1}. {step[2]}: (Jug A: {step[0]}L, Jug B: {step[1]}L) -> (Next State)")
    final_state = solution[-1]
    print(f"Final State: (Jug A: {final_state[0]}L, Jug B: {final_state[1]}L)")
else:
    print("No solution possible.")
```
